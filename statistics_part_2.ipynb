Q1- What is hypothesis testing in statistics?

Ans- Hypothesis testing in statistics is a method used to make decisions or inferences about a population based on sample data. It involves testing an assumption (called the null hypothesis) against an alternative hypothesis using statistical evidence.

Q2- What is the null hypothesis, and how does it differ from the alternative hypothesis?

Ans- The null hypothesis (H₀) is the assumption that there is no effect or no difference. The alternative hypothesis (H₁) is the claim that there is an effect or a difference.

Q3-  What is the significance level in hypothesis testing, and why is it important?

Ans- The significance level (α) is the probability of rejecting the null hypothesis when it is actually true, usually set at 0.05. It’s important because it defines the threshold for determining whether results are statistically significant.

Q4-  What does a P-value represent in hypothesis testing?

Ans- A p-value represents the probability of obtaining results as extreme as the observed ones, assuming the null hypothesis is true.

Q5-  How do you interpret the P-value in hypothesis testing?

Ans- If the p-value is less than the significance level (α), you reject the null hypothesis; otherwise, you fail to reject it.

Q6-  What are Type 1 and Type 2 errors in hypothesis testing?

Ans- Type 1 error is rejecting a true null hypothesis, while Type 2 error is failing to reject a false null hypothesis.

Q7-  What is the difference between a one-tailed and a two-tailed test in hypothesis testing?

Ans- A one-tailed test checks for an effect in one direction, while a two-tailed test checks for effects in both directions.

Q8- What is the Z-test, and when is it used in hypothesis testing?

Ans- The Z-test is a statistical test used to determine if there is a significant difference between sample and population means when the population standard deviation is known. It's used when the sample size is large (n > 30) or the population is normally distributed.

Q9-  How do you calculate the Z-score, and what does it represent in hypothesis testing?

Ans- The Z-score is calculated using the formula:

Z = (X - μ) / σ

Where X is the sample value, μ is the population mean, and σ is the population standard deviation. It represents how many standard deviations the sample value is away from the population mean.

Q10-  What is the T-distribution, and when should it be used instead of the normal distribution?

Ans- The T-distribution is a probability distribution used when the sample size is small (n < 30) and the population standard deviation is unknown. It should be used instead of the normal distribution in these cases to account for the increased variability in small samples.

Q11- What is the difference between a Z-test and a T-test?

Ans- A Z-test is used when the population standard deviation is known and the sample size is large, while a T-test is used when the population standard deviation is unknown and/or the sample size is small.

Q12-  What is the T-test, and how is it used in hypothesis testing?

Ans- The T-test is a statistical test used to compare the means of two groups or a sample mean with a population mean when the population standard deviation is unknown. It is used in hypothesis testing to determine if there is a significant difference between the groups or means.


q13- What is the relationship between Z-test and T-test in hypothesis testing?

Ans- The Z-test and T-test both compare means to test hypotheses, but the Z-test is used when the population standard deviation is known and the sample size is large, while the T-test is used when the population standard deviation is unknown or the sample size is small. As the sample size increases, the T-distribution approaches the normal distribution, making the T-test and Z-test similar.

Q14-  What is a confidence interval, and how is it used to interpret statistical results?

Ans- A confidence interval is a range of values that is used to estimate the true population parameter. It provides a range within which the parameter is likely to fall, with a certain level of confidence (e.g., 95%). It helps interpret statistical results by indicating the precision and reliability of the estimate.

Q15-  What is the margin of error, and how does it affect the confidence interval?

Ans- The margin of error is the amount of error allowed in a sample estimate, often based on the desired confidence level. It affects the confidence interval by determining its width; a larger margin of error results in a wider interval, indicating less precision in the estimate.

Q16-  How is Bayes' Theorem used in statistics, and what is its significance?

Ans- Bayes' Theorem is used to update the probability of a hypothesis based on new evidence. It combines prior knowledge (prior probability) with observed data (likelihood) to calculate the posterior probability. Its significance lies in its ability to refine predictions and make more informed decisions as new data becomes available.

Q17-  What is the Chi-square distribution, and when is it used?

Ans- The Chi-square distribution is a probability distribution used for hypothesis testing, particularly in tests of independence and goodness-of-fit. It is used when analyzing categorical data to determine if observed frequencies differ significantly from expected frequencies.

Q18- What is the Chi-square goodness of fit test, and how is it applied?

Ans- The Chi-square goodness of fit test is used to determine if a sample data matches an expected distribution. It compares the observed frequencies with the expected frequencies to assess if there are significant differences. It's applied by calculating the Chi-square statistic and comparing it to a critical value from the Chi-square distribution.

Q19-  What is the F-distribution, and when is it used in hypothesis testing?

Ans- The F-distribution is a probability distribution used to compare variances between two or more groups. It is commonly used in analysis of variance (ANOVA) and regression analysis to test if the group variances are significantly different.

Q20- What is an ANOVA test, and what are its assumptions?

Ans- An ANOVA (Analysis of Variance) test is used to compare the means of three or more groups to see if at least one group mean is significantly different from the others. The assumptions of ANOVA are:

1.Independence of observations
2.Normality of the data in each group
3.Homogeneity of variances (equal variances across groups).

Q21-  What are the different types of ANOVA tests?

Ans- The main types of ANOVA tests are:

1.One-way ANOVA: Compares the means of three or more groups based on one factor.
2.Two-way ANOVA: Compares the means of groups based on two factors, considering both individual and interaction effects.
3.Repeated measures ANOVA: Used when the same subjects are measured multiple times under different conditions.

Q22-  What is the F-test, and how does it relate to hypothesis testing?

Ans- The F-test is used to compare two variances or to test if the group means in ANOVA are significantly different. It relates to hypothesis testing by evaluating the ratio of variances (F-statistic) to determine if the observed differences are due to chance or if they are statistically significant.

```python
# Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results

import numpy as np
import scipy.stats as stats

# Sample data
sample_data = [12, 15, 14, 10, 13, 11, 16, 14, 13, 12]
sample_mean = np.mean(sample_data)
sample_size = len(sample_data)
sample_std = np.std(sample_data, ddof=1)

# Population data
population_mean = 14
population_std = 3  # Assume known population standard deviation

# Z-test calculation
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))
p_value = stats.norm.cdf(z_score)

# Results
print(f"Z-score: {z_score}")
print(f"P-value: {p_value}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the sample mean and the population mean.")
```

```python
# Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Function to calculate and visualize Type 1 and Type 2 errors
def visualize_type_errors(sample_size, population_mean, sample_mean, population_std, alpha=0.05):
    # Calculate Z-scores for critical values (rejection region)
    critical_value = stats.norm.ppf(1 - alpha / 2)  # Two-tailed test
    z_score_critical = critical_value

    # Calculate Z-scores for the sample
    z_sample = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate Type 1 Error (alpha) and Type 2 Error (beta)
    type_1_error = alpha
    type_2_error = stats.norm.cdf(-critical_value - (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))) + \
                    (1 - stats.norm.cdf(critical_value - (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))))

    # Results
    print(f"Sample Mean: {sample_mean}")
    print(f"Z-score for Sample: {z_sample}")
    print(f"Type 1 Error (alpha): {type_1_error}")
    print(f"Type 2 Error (beta): {type_2_error}")

    # Visualizing Type 1 and Type 2 errors
    x = np.linspace(-4, 4, 1000)  # Range of Z-scores
    y = stats.norm.pdf(x)  # Standard normal distribution

    plt.plot(x, y, label="Standard Normal Distribution")

    # Shade the rejection region (Type 1 Error)
    plt.fill_between(x, 0, y, where=(x <= -z_score_critical) | (x >= z_score_critical), color='red', alpha=0.5, label="Rejection Region (Type 1 Error)")

    # Shade the Type 2 Error region
    plt.fill_between(x, 0, y, where=(x >= -z_sample) & (x <= z_sample), color='yellow', alpha=0.5, label="Fail to Reject (Type 2 Error)")

    # Mark critical values and sample Z-score
    plt.axvline(-z_score_critical, color='green', linestyle='dashed', label=f"Critical Value (-{z_score_critical:.2f})")
    plt.axvline(z_score_critical, color='green', linestyle='dashed', label=f"Critical Value ({z_score_critical:.2f})")
    plt.axvline(z_sample, color='blue', linestyle='dashed', label=f"Sample Z-score: {z_sample:.2f}")

    # Adding labels and title
    plt.title("Type 1 and Type 2 Errors in Hypothesis Testing")
    plt.xlabel("Z-score")
    plt.ylabel("Probability Density")
    plt.legend()

    # Show the plot
    plt.show()

# Example Usage
visualize_type_errors(sample_size=100, population_mean=50, sample_mean=52, population_std=10)
```

```python
#Write a Python function to calculate the confidence interval for a sample mean and explain its significance

import numpy as np
import scipy.stats as stats

def confidence_interval(sample_data, confidence_level=0.95):
    """
    Function to calculate the confidence interval for a sample mean.

    Parameters:
    sample_data (list or np.array): The sample data
    confidence_level (float): The confidence level (default is 0.95 for 95% confidence)

    Returns:
    tuple: Lower and upper bounds of the confidence interval
    """
    # Calculate sample mean and sample standard deviation
    sample_mean = np.mean(sample_data)
    sample_std = np.std(sample_data, ddof=1)
    sample_size = len(sample_data)

    # Calculate the critical value for the confidence level
    alpha = 1 - confidence_level
    critical_value = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)  # t-distribution critical value

    # Calculate the margin of error
    margin_of_error = critical_value * (sample_std / np.sqrt(sample_size))

    # Calculate the confidence interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound

# Example Usage
sample_data = [23, 21, 24, 25, 22, 26, 27, 28, 22, 21]
confidence_level = 0.95
lower, upper = confidence_interval(sample_data, confidence_level)

print(f"Confidence Interval ({confidence_level*100}%): ({lower:.2f}, {upper:.2f})")
```

```python
# Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process

import numpy as np

def bayesian_inference(prior, likelihood, evidence):
    """
    Perform Bayesian Inference using Bayes' Theorem.

    Parameters:
    prior (float): The prior probability of the hypothesis.
    likelihood (float): The likelihood of the evidence given the hypothesis.
    evidence (float): The total probability of the evidence.

    Returns:
    float: The posterior probability of the hypothesis given the evidence.
    """
    # Bayes' Theorem
    posterior = (likelihood * prior) / evidence
    return posterior

# Example: Coin Bias Problem
# Prior: 50% chance the coin is biased, 50% chance it is fair
prior_biased = 0.5
prior_fair = 0.5

# Likelihood: If the coin is biased, the likelihood of getting 3 heads in 3 flips is high (e.g., 0.8)
likelihood_biased = 0.8
likelihood_fair = 0.125  # For a fair coin, the likelihood of 3 heads in 3 flips is (1/2)^3 = 0.125

# Evidence: The total probability of getting 3 heads in 3 flips, under all hypotheses
evidence = (likelihood_biased * prior_biased) + (likelihood_fair * prior_fair)

# Calculate posterior probabilities
posterior_biased = bayesian_inference(prior_biased, likelihood_biased, evidence)
posterior_fair = bayesian_inference(prior_fair, likelihood_fair, evidence)

print(f"Posterior probability of the coin being biased: {posterior_biased:.3f}")
print(f"Posterior probability of the coin being fair: {posterior_fair:.3f}")
```

```python
# Perform a Chi-square test for independence between two categorical variables in Python

import numpy as np
import scipy.stats as stats
import pandas as pd

# Example data: Number of people in different age groups and vehicle types
data = np.array([[30, 10, 20],  # Under 30: 30 Car, 10 Bike, 20 Bus
                 [40, 20, 10],  # 30-50: 40 Car, 20 Bike, 10 Bus
                 [10, 15, 25]]) # Over 50: 10 Car, 15 Bike, 25 Bus

# Create a DataFrame to represent the data
age_groups = ['Under 30', '30-50', 'Over 50']
vehicle_types = ['Car', 'Bike', 'Bus']
df = pd.DataFrame(data, columns=vehicle_types, index=age_groups)

# Perform the Chi-square test for independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(data)

# Display the results
print(f"Chi-square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")
print(f"Degrees of Freedom: {dof}")
print(f"Expected Frequencies Table: \n{expected}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("\nReject the null hypothesis: The variables (vehicle type and age group) are dependent.")
else:
    print("\nFail to reject the null hypothesis: The variables (vehicle type and age group) are independent.")
```

```python
# Write a Python program to calculate the expected frequencies for a Chi-square test based on observed data

import numpy as np

def calculate_expected_frequencies(observed_data):
    """
    Function to calculate expected frequencies for a Chi-square test.

    Parameters:
    observed_data (np.array): Observed frequency table (contingency table)

    Returns:
    np.array: Expected frequencies table
    """
    # Calculate the row sums and column sums
    row_totals = np.sum(observed_data, axis=1)
    column_totals = np.sum(observed_data, axis=0)

    # Total number of observations
    total = np.sum(observed_data)

    # Calculate expected frequencies
    expected_frequencies = np.outer(row_totals, column_totals) / total

    return expected_frequencies

# Example data: observed frequencies (rows: Age groups, columns: Vehicle types)
observed_data = np.array([[30, 10, 20],  # Under 30: 30 Car, 10 Bike, 20 Bus
                          [40, 20, 10],  # 30-50: 40 Car, 20 Bike, 10 Bus
                          [10, 15, 25]]) # Over 50: 10 Car, 15 Bike, 25 Bus

# Calculate the expected frequencies
expected_frequencies = calculate_expected_frequencies(observed_data)

# Display the expected frequencies
print("Observed Frequencies:\n", observed_data)
print("\nExpected Frequencies:\n", expected_frequencies)

```

```python
# Perform a goodness-of-fit test using Python to compare the observed data to an expected distribution

import numpy as np
import scipy.stats as stats

def goodness_of_fit_test(observed, expected):
    """
    Function to perform a Chi-square goodness-of-fit test.

    Parameters:
    observed (np.array): The observed frequencies
    expected (np.array): The expected frequencies

    Returns:
    tuple: Chi-square statistic, p-value, degrees of freedom
    """
    # Perform the Chi-square goodness-of-fit test
    chi2_stat, p_value = stats.chisquare(observed, expected)

    # Degrees of freedom
    dof = len(observed) - 1

    return chi2_stat, p_value, dof

# Example Data:
# Observed frequencies (e.g., a survey result of 100 people choosing a favorite color)
observed = np.array([25, 35, 20, 15, 5])  # Observed frequencies for colors: Red, Blue, Green, Yellow, Other

# Expected frequencies (e.g., expecting equal preference for each color)
expected = np.array([20, 20, 20, 20, 20])  # Expected frequencies: equal preference for each color

# Perform the Goodness-of-Fit Test
chi2_stat, p_value, dof = goodness_of_fit_test(observed, expected)

# Display the results
print(f"Chi-square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")
print(f"Degrees of Freedom: {dof}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("\nReject the null hypothesis: The observed frequencies do not fit the expected distribution.")
else:
    print("\nFail to reject the null hypothesis: The observed frequencies fit the expected distribution.")
```

```python
# Implement an F-test using Python to compare the variances of two random samples

import numpy as np
import scipy.stats as stats

def f_test(sample1, sample2):
    """
    Perform an F-test to compare the variances of two random samples.

    Parameters:
    sample1 (array): First sample data.
    sample2 (array): Second sample data.

    Returns:
    tuple: F-statistic, p-value, and decision based on significance level.
    """
    # Calculate sample variances
    var1 = np.var(sample1, ddof=1)  # Unbiased variance estimate
    var2 = np.var(sample2, ddof=1)

    # Calculate F-statistic: larger variance in the numerator
    if var1 > var2:
        F_statistic = var1 / var2
        df1 = len(sample1) - 1
        df2 = len(sample2) - 1
    else:
        F_statistic = var2 / var1
        df1 = len(sample2) - 1
        df2 = len(sample1) - 1

    # Perform the F-test
    p_value = 1 - stats.f.cdf(F_statistic, df1, df2)

    return F_statistic, p_value

# Example data: Two random samples
np.random.seed(42)
sample1 = np.random.normal(loc=50, scale=10, size=100)  # Sample 1 with mean 50 and std 10
sample2 = np.random.normal(loc=50, scale=20, size=100)  # Sample 2 with mean 50 and std 20

# Perform the F-test
F_statistic, p_value = f_test(sample1, sample2)

# Display the results
print(f"F-statistic: {F_statistic}")
print(f"P-value: {p_value}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("\nReject the null hypothesis: The variances are significantly different.")
else:
    print("\nFail to reject the null hypothesis: The variances are not significantly different.")
```

```python
# Write a Python program to perform an ANOVA test to compare means between multiple groups and interpret the results

import numpy as np
import scipy.stats as stats

def perform_anova(*groups):
    """
    Perform a one-way ANOVA test to compare the means of multiple groups.

    Parameters:
    *groups (arrays): The groups to compare.

    Returns:
    tuple: ANOVA F-statistic, p-value, and decision based on significance level.
    """
    # Perform the ANOVA test
    f_statistic, p_value = stats.f_oneway(*groups)

    return f_statistic, p_value

# Example data: Three groups of data
np.random.seed(42)
group1 = np.random.normal(loc=50, scale=10, size=30)  # Group 1 (mean = 50, std = 10)
group2 = np.random.normal(loc=55, scale=15, size=30)  # Group 2 (mean = 55, std = 15)
group3 = np.random.normal(loc=60, scale=20, size=30)  # Group 3 (mean = 60, std = 20)

# Perform the ANOVA test
f_statistic, p_value = perform_anova(group1, group2, group3)

# Display the results
print(f"ANOVA F-statistic: {f_statistic}")
print(f"P-value: {p_value}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("\nReject the null hypothesis: There is a significant difference between the group means.")
else:
    print("\nFail to reject the null hypothesis: There is no significant difference between the group means.")
```

```python
#Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def check_assumptions(*groups):
    """
    Check the assumptions for performing ANOVA: Normality, Independence, and Equal Variance.

    Parameters:
    *groups (arrays): The groups to check assumptions for.

    Returns:
    dict: A dictionary with results of assumption tests.
    """
    assumptions = {}

    # 1. Check for Normality using the Shapiro-Wilk test
    normality_results = []
    for i, group in enumerate(groups, start=1):
        stat, p_value = stats.shapiro(group)
        normality_results.append((stat, p_value))
        assumptions[f'Normality Group {i}'] = "Pass" if p_value > 0.05 else "Fail"

    # 2. Check for Equal Variance using Levene's Test
    stat, p_value = stats.levene(*groups)
    assumptions['Equal Variance'] = "Pass" if p_value > 0.05 else "Fail"

    # 3. Check for Independence (basic check through random sampling)
    # Assuming random sampling and no explicit test for independence; can check visually if required.
    assumptions['Independence'] = "Assumed to be random sampling"

    # Display the normality results
    for i, (stat, p_value) in enumerate(normality_results, start=1):
        print(f"Normality Test (Shapiro-Wilk) for Group {i}: Stat = {stat:.4f}, p-value = {p_value:.4f}")
        if p_value <= 0.05:
            print(f"Group {i} fails the normality assumption.")
        else:
            print(f"Group {i} passes the normality assumption.")

    # Display Levene's test results for equal variance
    print(f"\nEqual Variance Test (Levene's Test): Stat = {stat:.4f}, p-value = {p_value:.4f}")
    if p_value <= 0.05:
        print("The groups have significantly different variances.")
    else:
        print("The variances of the groups are equal.")

    # Assuming random sampling for independence
    print("\nIndependence Assumption: Assumed to be random sampling of groups.")

    return assumptions


# Example data: Three groups of data
np.random.seed(42)
group1 = np.random.normal(loc=50, scale=10, size=30)  # Group 1 (mean = 50, std = 10)
group2 = np.random.normal(loc=55, scale=15, size=30)  # Group 2 (mean = 55, std = 15)
group3 = np.random.normal(loc=60, scale=20, size=30)  # Group 3 (mean = 60, std = 20)

# Check assumptions
assumptions = check_assumptions(group1, group2, group3)

# Print the assumption check results
print("\nAssumptions Check Results:")
for assumption, result in assumptions.items():
    print(f"{assumption}: {result}")
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.graphics.factorplots import interaction_plot

# 1. Simulate sample data
np.random.seed(0)
n = 30  # number of samples per group
treatment = np.repeat(['A', 'B'], n * 2)
gender = np.tile(np.repeat(['Male', 'Female'], n), 2)

# Generate random scores with some main and interaction effects
score = (
    50 +
    (np.array(treatment) == 'B') * 5 +
    (np.array(gender) == 'Female') * 7 +
    ((np.array(treatment) == 'B') & (np.array(gender) == 'Female')) * 3 +
    np.random.normal(0, 5, n * 4)
)

# Create DataFrame
data = pd.DataFrame({
    'Treatment': treatment,
    'Gender': gender,
    'Score': score
})

# 2. Fit the two-way ANOVA model with interaction
model = ols('Score ~ C(Treatment) * C(Gender)', data=data).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# 3. Interpret results
print("\nInterpretation:")
for effect, row in anova_table.iterrows():
    p = row['PR(>F)']
    if p < 0.05:
        print(f"✅ Significant effect: {effect} (p = {p:.4f})")
    else:
        print(f"❌ No significant effect: {effect} (p = {p:.4f})")

# 4. Visualize results

# Boxplot by Treatment and Gender
plt.figure(figsize=(8, 6))
sns.boxplot(x='Treatment', y='Score', hue='Gender', data=data)
plt.title("Boxplot of Scores by Treatment and Gender")
plt.grid(True)
plt.show()

# Interaction plot
plt.figure(figsize=(8, 6))
interaction_plot(data['Treatment'], data['Gender'], data['Score'],
                 markers=['D', '^'], colors=['blue', 'green'])
plt.title("Interaction Plot: Treatment vs Gender on Score")
plt.xlabel("Treatment")
plt.ylabel("Mean Score")
plt.grid(True)
plt.show()
```

```python
#Write a Python program to visualize the F-distribution and discuss its use in hypothesis testing

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

# Parameters for F-distribution (degrees of freedom)
dof1 = 5   # Numerator degrees of freedom
dof2 = 20  # Denominator degrees of freedom

# Generate x values and corresponding F-distribution PDF
x = np.linspace(0, 5, 500)
pdf = f.pdf(x, dfn=dof1, dfd=dof2)

# Critical value for alpha = 0.05 (right-tailed test)
alpha = 0.05
f_critical = f.ppf(1 - alpha, dfn=dof1, dfd=dof2)

# Plotting the F-distribution
plt.figure(figsize=(10, 6))
plt.plot(x, pdf, label=f'F-distribution (df1={dof1}, df2={dof2})', color='blue')
plt.axvline(f_critical, color='red', linestyle='--', label=f'Critical value (α = {alpha}) ≈ {f_critical:.2f}')

# Shade the rejection region
x_fill = np.linspace(f_critical, x.max(), 200)
plt.fill_between(x_fill, f.pdf(x_fill, dfn=dof1, dfd=dof2), color='red', alpha=0.3, label='Rejection Region')

# Labels and legend
plt.title('F-distribution and Critical Region (α = 0.05)')
plt.xlabel('F-statistic')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results

import numpy as np
from scipy.stats import chi2

# 1. Simulate sample data
np.random.seed(42)
sample = np.random.normal(loc=50, scale=8, size=30)  # true std = 8

# 2. Hypothesized population variance
sigma_0_squared = 64  # e.g., hypothesized std = 8

# 3. Calculate test statistic
n = len(sample)
sample_variance = np.var(sample, ddof=1)
chi_square_stat = (n - 1) * sample_variance / sigma_0_squared

# 4. Compute critical values and p-value
alpha = 0.05
df = n - 1
chi2_lower = chi2.ppf(alpha / 2, df)
chi2_upper = chi2.ppf(1 - alpha / 2, df)
p_val = 2 * min(chi2.cdf(chi_square_stat, df), 1 - chi2.cdf(chi_square_stat, df))

# 5. Output results
print(f"Sample Variance: {sample_variance:.2f}")
print(f"Chi-square Statistic: {chi_square_stat:.4f}")
print(f"Degrees of Freedom: {df}")
print(f"Critical Values (α={alpha}): Lower = {chi2_lower:.2f}, Upper = {chi2_upper:.2f}")
print(f"P-value: {p_val:.4f}")

# 6. Decision
if chi_square_stat < chi2_lower or chi_square_stat > chi2_upper:
    print("✅ Reject H0: Variance is significantly different from hypothesized value.")
else:
    print("❌ Fail to reject H0: No significant difference in variance.")
```

```python
# Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f

# 1. Simulate data for two groups
np.random.seed(42)
group1 = np.random.normal(loc=50, scale=5, size=30)   # Variance = 25
group2 = np.random.normal(loc=50, scale=10, size=30)  # Variance = 100

# 2. Calculate sample variances
var1 = np.var(group1, ddof=1)
var2 = np.var(group2, ddof=1)

# 3. Compute F-statistic (larger variance / smaller variance)
if var1 > var2:
    F = var1 / var2
    dfn = len(group1) - 1
    dfd = len(group2) - 1
else:
    F = var2 / var1
    dfn = len(group2) - 1
    dfd = len(group1) - 1

# 4. Significance level and critical values
alpha = 0.05
F_critical_upper = f.ppf(1 - alpha / 2, dfn, dfd)
F_critical_lower = f.ppf(alpha / 2, dfn, dfd)

# 5. P-value
p_value = 2 * min(f.cdf(F, dfn, dfd), 1 - f.cdf(F, dfn, dfd))

# 6. Print results
print(f"Sample Variances: Group1 = {var1:.2f}, Group2 = {var2:.2f}")
print(f"F-statistic: {F:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Critical Values (two-tailed, α=0.05): [{F_critical_lower:.2f}, {F_critical_upper:.2f}]")

if F < F_critical_lower or F > F_critical_upper:
    print("✅ Reject H0: Variances are significantly different.")
else:
    print("❌ Fail to reject H0: No significant difference in variances.")

# 7. Visualize F-distribution
x = np.linspace(0.1, 5, 500)
y = f.pdf(x, dfn, dfd)

plt.figure(figsize=(10, 6))
plt.plot(x, y, label=f'F-distribution (dfn={dfn}, dfd={dfd})')
plt.axvline(F, color='green', linestyle='--', label=f'F-statistic = {F:.2f}')
plt.axvline(F_critical_upper, color='red', linestyle='--', label='Upper Critical Value')
plt.axvline(F_critical_lower, color='red', linestyle='--', label='Lower Critical Value')

# Shade rejection regions
x_fill_right = np.linspace(F_critical_upper, x.max(), 100)
plt.fill_between(x_fill_right, f.pdf(x_fill_right, dfn, dfd), color='red', alpha=0.3)

x_fill_left = np.linspace(x.min(), F_critical_lower, 100)
plt.fill_between(x_fill_left, f.pdf(x_fill_left, dfn, dfd), color='red', alpha=0.3)

plt.title('F-Test for Variance Comparison')
plt.xlabel('F value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
